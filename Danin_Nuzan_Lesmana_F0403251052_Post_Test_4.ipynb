```python
class Komoditas:
    """Merepresentasikan data dan pengelolaan suatu komoditas pertanian."""

    def __init__(self, nama, jenis, harga_per_kg, stok):
        """Membuat objek Komoditas dengan nama, jenis, harga, dan stok."""
        self.nama = nama
        self.jenis = jenis
        self.harga_per_kg = harga_per_kg
        self.stok = stok

    def hitung_nilai_stok(self):
        """Menghitung total nilai stok berdasarkan harga per kg."""
        return self.harga_per_kg * self.stok

    def tambah_stok(self, jumlah):
        """Menambahkan jumlah stok komoditas."""
        self.stok += jumlah


class Gudang:
    """Merepresentasikan gudang yang menyimpan berbagai komoditas."""

    def __init__(self, nama_gudang, kapasitas):
        """Membuat objek Gudang dengan nama dan kapasitas penyimpanan."""
        self.nama_gudang = nama_gudang
        self.kapasitas = kapasitas
        self.daftar_komoditas = []

    def tambah_komoditas(self, komoditas):
        """Menambahkan objek komoditas ke dalam gudang."""
        self.daftar_komoditas.append(komoditas)

    def tampilkan_stok(self):
        """Menampilkan seluruh komoditas yang tersimpan di gudang."""
        print(f"Gudang: {self.nama_gudang}")
        for komoditas in self.daftar_komoditas:
            print(f"- {komoditas.nama}: {komoditas.stok} kg")


# Membuat objek
jagung = Komoditas("Jagung", "Serealia", 7000, 100)
kedelai = Komoditas("Kedelai", "Kacang-kacangan", 12000, 50)

gudang = Gudang("Gudang Agro Makmur", 1000)

# Menambahkan komoditas ke gudang
gudang.tambah_komoditas(jagung)
gudang.tambah_komoditas(kedelai)

# Menampilkan data
gudang.tampilkan_stok()

print("Nilai stok jagung:", jagung.hitung_nilai_stok())
```

```python
class Komoditas:
    """Merepresentasikan data dan pengelolaan suatu komoditas pertanian."""

    def __init__(self, nama, jenis, harga_per_kg, stok):
        """Membuat objek Komoditas dengan validasi harga dan stok."""

        if harga_per_kg < 0:
            raise ValueError("Harga tidak boleh negatif!")

        if stok < 0:
            raise ValueError("Stok tidak boleh negatif!")

        self.nama = nama
        self.jenis = jenis
        self.harga_per_kg = harga_per_kg
        self.stok = stok

    def hitung_nilai_stok(self):
        """Menghitung dan mengembalikan nilai total stok."""
        return self.harga_per_kg * self.stok

    def tambah_stok(self, jumlah):
        """Menambahkan jumlah stok komoditas."""
        if jumlah < 0:
            raise ValueError("Jumlah penambahan stok tidak boleh negatif!")
        self.stok += jumlah

    def __str__(self):
        """Mengembalikan informasi objek Komoditas dalam bentuk teks."""
        return f"{self.nama} - Stok: {self.stok} kg - Harga: Rp{self.harga_per_kg}"
```

```python
class Komoditas:
    """Merepresentasikan data dan pengelolaan komoditas pertanian."""

    jumlah_objek = 0  # Atribut kelas

    def __init__(self, nama, jenis, harga_per_kg, stok):
        """Membuat objek Komoditas dan menambah jumlah objek."""

        if harga_per_kg < 0:
            raise ValueError("Harga tidak boleh negatif!")

        if stok < 0:
            raise ValueError("Stok tidak boleh negatif!")

        self.nama = nama
        self.jenis = jenis
        self.harga_per_kg = harga_per_kg
        self.stok = stok

        # Menambah jumlah objek setiap kali objek dibuat
        Komoditas.jumlah_objek += 1

    def hitung_nilai_stok(self):
        """Menghitung nilai total stok."""
        return self.harga_per_kg * self.stok

    def tambah_stok(self, jumlah):
        """Menambahkan jumlah stok."""
        self.stok += jumlah

    def __str__(self):
        """Menampilkan informasi objek."""
        return f"{self.nama} - Stok: {self.stok} kg"
```

```python
jagung = Komoditas("Jagung", "Serealia", 7000, 100)
print("Jumlah objek:", Komoditas.jumlah_objek)

kedelai = Komoditas("Kedelai", "Kacang-kacangan", 12000, 50)
print("Jumlah objek:", Komoditas.jumlah_objek)

beras = Komoditas("Beras", "Serealia", 14000, 80)
print("Jumlah objek:", Komoditas.jumlah_objek)
```

```python
class Komoditas:
    """Merepresentasikan komoditas pertanian."""

    diskon = 0.10  # Diskon 10% untuk semua objek

    def __init__(self, nama, harga_per_kg):
        """Membuat objek Komoditas."""
        self.nama = nama
        self.harga_per_kg = harga_per_kg

    def harga_setelah_diskon(self):
        """Menghitung harga setelah diskon kelas."""
        return self.harga_per_kg * (1 - Komoditas.diskon)


jagung = Komoditas("Jagung", 7000)
kedelai = Komoditas("Kedelai", 12000)

print(jagung.harga_setelah_diskon())
print(kedelai.harga_setelah_diskon())

# Mengubah atribut kelas
Komoditas.diskon = 0.20

print("\nSetelah diskon diubah menjadi 20%:")
print(jagung.harga_setelah_diskon())
print(kedelai.harga_setelah_diskon())
```

```python
class MesinSortasi:
    """Merepresentasikan mesin sortasi komoditas pertanian."""

    def __init__(self, nama, kapasitas_maks, suhu_maks):
        """Membuat objek mesin dengan batas kapasitas dan suhu maksimum."""
        self.nama = nama
        self.kapasitas_maks = kapasitas_maks
        self.suhu_maks = suhu_maks

    def proses_sortasi(self, jumlah, suhu):
        """Menjalankan sortasi jika tidak melewati batas fisik mesin."""

        if jumlah > self.kapasitas_maks:
            raise ValueError(
                f"Kapasitas melebihi batas! Maksimal {self.kapasitas_maks} kg."
            )

        if suhu > self.suhu_maks:
            raise ValueError(
                f"Suhu melebihi batas! Maksimal {self.suhu_maks} °C."
            )

        print(f"Sortasi berhasil: {jumlah} kg pada suhu {suhu} °C.")


# Membuat objek mesin
mesin = MesinSortasi("Mesin Sortasi Jagung", 500, 40)

# Uji input valid
try:
    mesin.proses_sortasi(300, 30)
except ValueError as e:
    print("Error:", e)

# Uji kapasitas yang melanggar batas
try:
    mesin.proses_sortasi(600, 30)
except ValueError as e:
    print("Error:", e)

# Uji suhu yang melanggar batas
try:
    mesin.proses_sortasi(300, 50)
except ValueError as e:
    print("Error:", e)
```
