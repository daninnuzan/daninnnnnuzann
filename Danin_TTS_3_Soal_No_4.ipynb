```python
# Fungsi 1: menambahkan nama petani ke List
def tambah_petani(daftar_petani, nama):
    daftar_petani.append(nama)


# Fungsi 2: menyimpan identitas petani dalam Dictionary
def simpan_identitas(kode_anggota, tanggal_gabung):
    identitas = {
        "kode_anggota": kode_anggota,
        "tanggal_gabung": tanggal_gabung
    }
    return identitas


# Fungsi 3: menyimpan detail transaksi dalam Dictionary
def buat_transaksi(komoditas, jumlah, harga_per_kg):
    transaksi = {
        "komoditas": komoditas,
        "jumlah_kg": jumlah,
        "harga_per_kg": harga_per_kg,
        "total": jumlah * harga_per_kg
    }
    return transaksi


# Fungsi 4: menambahkan komoditas unik ke Set
def tambah_komoditas(daftar_komoditas, komoditas):
    daftar_komoditas.add(komoditas)


# Fungsi untuk menampilkan seluruh data
def tampilkan_data(daftar_petani, identitas, transaksi, daftar_komoditas):
    print("=== DATA KOPERASI TANI ===")
    print("Daftar Petani:", daftar_petani)
    print("Identitas Petani:", identitas)
    print("Detail Transaksi:", transaksi)
    print("Komoditas Unik:", daftar_komoditas)


# Fungsi utama
def main():
    # List nama petani
    daftar_petani = []

    # Menambahkan nama petani
    tambah_petani(daftar_petani, "Budi")
    tambah_petani(daftar_petani, "Siti")
    tambah_petani(daftar_petani, "Andi")

    # Dictionary identitas petani
    identitas = simpan_identitas("KT001", "12-08-2026")

    # Dictionary detail transaksi
    transaksi = buat_transaksi("Padi", 100, 7000)

    # Set komoditas unik
    daftar_komoditas = set()
    tambah_komoditas(daftar_komoditas, "Padi")
    tambah_komoditas(daftar_komoditas, "Jagung")
    tambah_komoditas(daftar_komoditas, "Padi")

    # Menampilkan data
    tampilkan_data(
        daftar_petani,
        identitas,
        transaksi,
        daftar_komoditas
    )


# Menjalankan program
main()
```
